```python
# 1. PROGRAM TO GENERATE POSITIVE NUMBERS FROM A LIST
n = int(input("Enter number of elements: "))
a = []
for i in range(n):
    x = int(input("Enter a number: "))
    a.append(x)
positive = [x for x in a if x > 0]
print("Positive numbers:", positive)
```

```python
# 2. PROGRAM TO FIND THE SQUARE OF N NUMBERS
n = int(input("Enter N: "))
a = []
for i in range(n):
    x = int(input("Enter a number: "))
    a.append(x)
square = [x*x for x in a]
print("Squares:", square)
```
