```python
#EXCHANGE FIRST AND LAST CHARACTERS
s=input("Enter a new string:")
new=s[-1]+s[1:-1]+s[0]
print("New string=",new)
```

```python
#FIRST 2 AND LAST 2 CHARACTERS
S=input("Enter a new string:")
new=s[:2]+s[-2:]
print("New string=",new)
```

```python
#SWAP CHARACTER AT POSITION 1
s1=input("Enter first string:")
s2=input("Enter second string;")
new1=s2[0]+s1[1:]
new2=s1[0]+s2[1:]
print("New string=",new1,new2)
```

```python
#REPLACE OCCURRENCES OF FIRST CHARACTER WITH $
s=input("Enter a new string:")
first=s[0]
new=first+s[1:].replace(first,"$")
print("New string=",new)
```
