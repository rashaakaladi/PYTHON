```python
#LIST OF COLORS- DISPLAY FIRST AND LAST
colors=input("Enter colors separated by comma:").split(",")
print("First color=",colors[0])
print("Last color=",colors[-1])
```

```python
#FIND THE EXTENSION OF A FILENAME
filename=input("Enter filename:")
extension=filename.split(".")[-1]
print("Extension=",extension)
```

```python
#COMMA-SEPARATED NUMBERS AND FIND THEIR SUM
numbers=input("Enter numbers separated by comma:").split(",")
numbers=[int(x)for x in numbers]
print("List=",numbers)
print("Sum=",sum(numbers))

```
