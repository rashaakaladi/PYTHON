```python
#PRINT OUT ALL COLORS FROM COLOR SET1 NOT CONTAINED IN COLOR SET2
set1=set(input("Enter colors in set 1:").split(","))
set2=set(input("Enter colors in set 2:").split(","))
result=set1-set2
print("Colors in set 1 but not in set 2=",result)
```

```python
#PRINT OUT ALL COLORS FROM COLOR SET1 AND COLOR SET2
set1=set(input("Enter colors in set 1:").split(","))
set2=set(input("Enter colors in set 2:").split(","))
result=set1|set2
print("Colors in set 1 and set 2=",result)
```
