```python
# 1. PROGRAM TO FIND THE LENGTH OF THE LARGEST WORD
words = input("Enter words separated by space:").split()
length = max(len(word) for word in words)
longest = [word for word in words if len(word) == length]
print("Longest words:", longest)
print("Length:", length)
```

```python
# 2. PROGRAM TO PRINT A STAR PATTERN
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
for i in range(4, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# 3. PROGRAM TO FIND ALL FACTORS OF A NUMBER
n = int(input("Enter a number:" ))
print("Factors are:")
for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
```

```python
# 4. PROGRAM TO FIND AREAS USING LAMBDA FUNCTIONS
square = lambda a: a * a
rectangle = lambda l, b: l * b
triangle = lambda b, h: 0.5 * b * h

a = float(input("Enter side of square: "))
print("Area of square:", square(a))

l = float(input("Enter length of rectangle: "))
b = float(input("Enter breadth of rectangle: "))
print("Area of rectangle:", rectangle(l,b))

b = float(input("Enter base of triangle: "))
h = float(input("Enter height of triangle: "))
print("Area of triangle:", triangle(b,h))

```
