```python
import os
os.makedirs("graphics/graphics3D", exist_ok=True)

open("graphics/__init__.py", "w").close()
open("graphics/graphics3D/__init__.py","w").close() 
```

```python
%%writefile graphics/rectangle.py

def area(l, b):
    return l * b

def perimeter(l, b):
    return 2 * (l + b)
```

```python
%%writefile graphics/circle.py

import math

def area(r):
    return math.pi * r * r

def perimeter(r):
    return 2 * math.pi * r
```

```python
%%writefile graphics/graphics3D/cuboid.py

def area(l, b, h):
    return 2 * (l*b + l*h + b*h)

def perimeter(l, b, h):
    return 4 * (l + b + h)
```

```python
%%writefile graphics/graphics3D/sphere.py

import math

def area(r):
    return 4 * math.pi * r * r

def perimeter(r):
    return 2 * math.pi * r
```

```python
import graphics.rectangle as rect
from graphics import circle
from graphics.graphics3D import cuboid
from graphics.graphics3D.sphere import *

while True:
    print("\n1. Rectangle")
    print("2. Circle")
    print("3. Cuboid")
    print("4. Sphere")
    print("5. Exit")

    ch = int(input("Enter your choice: "))

    if ch == 1:
        l = int(input("Enter length: "))
        b = int(input("Enter breadth: "))
        print("Area:", rect.area(l, b))
        print("Perimeter:", rect.perimeter(l,b))

    elif ch == 2:
        r = int(input("Enter radius: "))
        print("Area:",round(circle.area(r),2))
        print("Perimeter:", round(circle.perimeter(r), 2))

    elif ch == 3:
        l = int(input("Enter length: "))
        b = int(input("Enter breadth: "))
        h = int(input("Enter height: "))
        print("Total surface area:", cuboid.area(l, b, h))
        print("Perimeter:", cuboid.perimeter(l, b, h))

    elif ch == 4:
        r = int(input("Enter radius: "))
        print("Area:", round(area(r), 2))
        print("Perimeter:", round(perimeter(r), 2))

    elif ch == 5:
        break

    else:
        print("Invalid choice")
```
