```python
# ADD "ing" OR "ly"
string=input("Enter a string:")
if string.endswith("ing"):
    result = string + "ly"
else:
    result = string + "ing"
print("Result=",result)
```

```python
# CHECK WHETHER A STRING IS PALINDROME OR NOT
string=input("Enter a string:")
if string == string[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```
