```python
#GET THE 3RD ELEMENT AND 3RD ELEMENT FROM LAST OF A TUPLE
t=tuple(input("Enter elements separated by comma:").split(","))
print("3rd element =",t[2])
print("3rd element from last =",t[-3])
```

```python
#ACCEPT NUMBERS,CREATE TUPLE AND FIND AVERAGE
numbers=input("Enter numbers seperated by comma:")
numbers=numbers.split(",")
a=int(numbers[0])
b=int(numbers[1])
c=int(numbers[2])
numbers=(a,b,c)
average=sum(numbers)/len(numbers)
print("Tuple=",numbers)
print("Average=",average)
                      
```
