```python
%%writefile rectangle.py

def area(l, b):
    return l * b

def perimeter(l, b):
    return 2 * (l + b)
```

```python
%%writefile circle.py
import math

def area(r):
    return math.pi * r * r

def perimeter(r):
    return 2 * math.pi * r
```

```python
import rectangle
import circle

while True:
    print("1.Rectangle")
    print("2.Circle")
    print("3.Exit")
    
    ch = int(input("Enter your choice: "))
    
    if ch == 1:
        l = int(input("Enter length: "))
        b = int(input("Enter breadth: "))
        print("Rectangle area:", rectangle.area(l, b))
        print("Rectangle perimeter:", rectangle.perimeter(l, b))
        
    elif ch == 2:
        r = int(input("Enter radius: "))
        print("Circle area:", round(circle.area(r), 2))
        print("Circle perimeter:", round(circle.perimeter(r), 2))
        
    elif ch == 3:
        break
        
    else:
        print("Invalid choice")
```
