```python
#ACCEPT 5 STUDENTS AND SORT BY ROLL NUMBER,(ROLL NUMBER=KEY,NAME=VALUE)
students={}
for i in range(5):
    rollno=int(input("Enter roll number:"))
    name=input("Enter student name:")
    students[rollno]=name
print("Ascending order=",dict(sorted(students.items())))
print("Descending order=",dict(sorted(students.items(),reverse=True)))
```

```python
#MERGE TWO DICTIONARIES USING ,(update())
dic1={}
for i in range (3):
    rollno=int(input("Enter roll number for dic1:"))
    name=input("Enter name for dic1:")
    dic1[rollno]=name
dic2={}
for i in range (2):
    rollno=int(input("Enter roll number for dic2:"))
    name=input("Enter name for dic2:")
    dic2[rollno]=name
dic1.update(dic2)
print("Merged dictionary=",dic1)
```
