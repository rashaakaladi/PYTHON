```python
# 1. RECTANGLE - AREA, PERIMETER AND COMPARISON
class Rectangle:
    def get(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        return self.l * self.b

    def perimeter(self):
        return 2 * (self.l + self.b)

r1 = Rectangle()
r2 = Rectangle()

l1 = int(input("Enter length of Rectangle 1: "))
b1 = int(input("Enter breadth of Rectangle 1: "))

l2 = int(input("Enter length of Rectangle 2: "))
b2 = int(input("Enter breadth of Rectangle 2: "))

r1.get(l1, b1)
r2.get(l2, b2)

p = r1.area()
q = r2.area()

print("Area of Rectangle 1:", p)
print("Perimeter of Rectangle 1:", r1.perimeter())

print("Area of Rectangle 2:", q)
print("Perimeter of Rectangle 2:", r2.perimeter())

if p > q:
    print("Rectangle 1 is greater")
elif p < q:
    print("Rectangle 2 is greater")
else:
    print("Both Rectangles are equal")
```

```python
# 2. BANK ACCOUNT – DEPOSIT AND WITHDRAW
class BankAccount:
    def __init__(self, a_name, a_no, a_type, a_balance):
        self.acc_name = a_name
        self.acc_no = a_no
        self.acc_type = a_type
        self.acc_balance = a_balance

    def deposit(self, a_deposit):
        print("Initial balance:", self.acc_balance)
        self.acc_balance += a_deposit
        print(a_deposit, "is deposited")
        print("Current balance:", self.acc_balance)

    def withdraw(self):
        amount = int(input("Withdraw amount: "))
        print("Initial balance:", self.acc_balance)

        if amount > self.acc_balance:
            print("You don't have enough balance to withdraw!")
        else:
            print(amount, "is withdrawn")
            self.acc_balance -= amount

        print("Current balance:", self.acc_balance)

    def acc_info(self):
        print("Account holder Name:", self.acc_name)
        print("Account Number:", self.acc_no)
        print("Account Type:", self.acc_type)
        print("Account Balance:", self.acc_balance)

name = input("Enter name: ")
no = int(input("Enter account number: "))
acc_type = input("Enter account type: ")
bal = int(input("Enter initial balance: "))

holder = BankAccount(name, no, acc_type, bal)

while True:
    ch = int(input(
        "\n1. Deposit 2. Withdraw 3. Account info 4. Exit\n"
        "Choose your option: "))
    
    if ch == 1:
        amount = int(input("Deposit amount: "))
        holder.deposit(amount)

    elif ch == 2:
        holder.withdraw()

    elif ch == 3:
        holder.acc_info()

    elif ch == 4:
        print("Exiting...")
        break

    else:
        print("Invalid choice!")
```

```python
# 3. MULTILEVEL INHERITANCE - PUBLISHER, BOOK AND AND PYTHON
class Publisher:
    def __init__(self, pname):
        self.pubname = pname

    def display1(self):
        print("Book details")
        print("Publisher name =", self.pubname)

class Book(Publisher):
    def __init__(self, pname, bname, author):
        Publisher.__init__(self, pname)
        self.title = bname
        self.auth = author

    def display2(self):
        print("Title =", self.title)
        print("Author =", self.auth)

class Python(Book):
    def __init__(self, pname, title, au, price, np):
        Book.__init__(self, pname, title, au)
        self.rate = price
        self.nop = np

    def display3(self):
        print("Rate =", self.rate)
        print("No. of pages =", self.nop)

p = input("Enter publisher name: ")
b = input("Enter book name: ")
a = input("Enter author name: ")
r = int(input("Enter rate of book: "))
n = int(input("Enter no. of pages: "))

obj = Python(p, b, a, r, n)

obj.display1()
obj.display2()
obj.display3()
```

```python
# 4. OPERATOR OVERLOADING - COMPARE TWO RECTANGLES USING '<'
class rectangle:
    def __init__(self, l, w):
        self.__length = l
        self.__width = w
        self.a = l * w

    def __lt__(self, other):
        if self.a < other.a:
            print("rectangle 2 is greater")
        else:
            print("rectangle 1 is greater")

l1 = int(input("Enter length of rectangle1: "))
w1 = int(input("Enter width of rectangle1: "))
l2 = int(input("Enter length of rectangle2: "))
w2 = int(input("Enter width of rectangle2: "))

r1 = rectangle(l1, w1)
r2 = rectangle(l2, w2)

r1 < r2
```

```python
# 5. OPERATOR OVERLOADING - ADDITION OF TWO TIME OBJECTS
class Time:
    def __init__(self, hr, mint, sec):
        self.__h = hr
        self.__m = mint
        self.__s = sec

    def __add__(self, other):
        hh = self.__h + other.__h
        mm = self.__m + other.__m
        ss = self.__s + other.__s

        if ss >= 60:
            m = int(ss / 60)
            ss = ss % 60
            mm += m

        if mm >= 60:
            h = int(mm / 60)
            mm = mm % 60
            hh += h

        print("Added time..")
        print("Hour:", hh, "Minute:", mm, "Second:", ss)

h1 = int(input("Enter hour1: "))
m1 = int(input("Enter minute1: "))
s1 = int(input("Enter second1: "))

h2 = int(input("Enter hour2: "))
m2 = int(input("Enter minute2: "))
s2 = int(input("Enter second2: "))

t1 = Time(h1, m1, s1)
t2 = Time(h2, m2, s2)

t1 + t2
```
