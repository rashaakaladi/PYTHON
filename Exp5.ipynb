```python
#GET THE 3RD ELEMENT AND 3RD ELEMENT FROM LAST OF A TUPLE
t=(10,20,30,40,50)
print("3rd element =",t[2])
print("3rd element from last =",t[-3])
```

```python
#ACCEPT COMMA-SEPARATED NUMBERS,CREATE A TUPLE AND FIND AVERAGE OF ALL
numbers=("Enter numbers separated by comma:")
numbers=numbers.split(",")
a=int(numbers[0])
b=int(numbers[1])
c=int(numbers[2])
numbers=(a,b,c)
average=sum(numbers)/len(numbers)
print("Tuple=",numbers)
print("Average=",average)
```
