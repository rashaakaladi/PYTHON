```python
# 1. COUNT THE OCCURRENCES OF EACH WORD
text=input("Enter a line of text:")
words=text.split()
count={}
for word in words:
    if word in count:
        count[word]=count[word]+1
    else:
        count[word]=1
for word in count:
 print(word,"=",count[word])
```

```python
# 2. FOR ALL VALUES GREATER THAN 100, STORE 'OVER' INSTEAD
numbers=input("Enter numbers separated by comma:").split(",")
result=[]
for x in numbers:
    x=int(x)
    if x>100:
        result.append("over")
    else:
        result.append(x)
print("Result=",result)
```

```python
# 3. COUNT THE OCCURRENCES OF 'A' WITHIN A LIST
names=input("Enter first names separated by comma:").split(",")
count = 0
for name in names:
    for ch in name:
        if ch =='a':
            count = count+1
print("Number of 'a'=",count)
```

```python
# 4. GCD OF TWO NUMBERS
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
while b != 0:
    r = a%b
    a = b
    b = r
print("GCD=",a)    
```

```python
# 5. REMOVE EVEN NUMBERS FROM A LIST
numbers=input("Enter numbers separated by comma:").split(",")
result=[]
for x in numbers:
    x=int(x)
    if x%2 != 0:
        result.append(x)
print("List after removing even numbers=",result)
```

```python
# 6. FIND FACTORIAL OF A NUMBER
n=int(input("Enter a number:"))
fact=1
for i in range(1,n+1):
    fact=fact*i
print("Factorial=",fact)
```

```python
# 7. GENERATE FIBONACCI SERIES OF N TERMS
n=int(input("Enter number of terms:"))
a=0
b=1
print("Fibonacci series:")
for i in range(n):
    print(a,end=" ")
    c=a+b
    a=b
    b=c
```

```python
# 8. FIND THE SUM OF ALL ITEMS IN A LIST
numbers=input("Enter numbers separated by comma:").split(",")
sum=0
for x in numbers:
    sum=sum+int(x)
print("Sum =",sum)
```

```python
# 9. FOUR-DIGIT NUMBERS WITH ALL DIGITS EVEN AND PERFECT SQUARE
start=int(input("Enter starting value:"))
end=int(input("Enter ending value:"))
for n in range(start,end + 1):
    if n >= 1000 and n <= 9999:
        a = n // 1000
        b = (n // 100) % 10
        c = (n // 10) % 10
        d = n % 10
        if a % 2 == 0 and b % 2 == 0 and c % 2 == 0 and d % 2 == 0:
            for i in range(1,100):
                if i * i == n:
                    print(n)
                    break
   
```

```python
# 10. COUNT THE NUMBER OF CHARACTERS IN A STRING
text = input("Enter a string: ")
count = {}

for ch in text:
    if ch in count:
        count[ch] = count[ch] + 1
    else:
        count[ch] = 1

for ch in count:
    print(ch, "=", count[ch])
```

```python
# 11. COMPARE TWO LISTS
del sum
list1 = input("Enter first list: ").split(",")
list2 = input("Enter second list: ").split(",")

for i in range(len(list1)):
    list1[i] = int(list1[i])

for i in range(len(list2)):
    list2[i] = int(list2[i])

if sum(list1) == sum(list2):
    print("Both lists have same sum")
else:
    print("Lists have different sums")

common = False

for x in list1:
    for y in list2:
        if x == y:
            common = True
            print("Common value =", x)

if common == False:
    print("No common value")
```
